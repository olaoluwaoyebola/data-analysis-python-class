# **Introduction to Numpy**

NumPy (Numerical Python) is a fundamental library for scientific computing in Python. It provides:

1. A powerful N-dimensional array object

2. Sophisticated broadcasting functions

3. Tools for integrating C/C++ code

4. Linear algebra, Fourier transform, and random number capabilities


## **Why NumPy?**

1. Performance: NumPy arrays are much faster than Python lists for numerical operations

2. Memory Efficiency: NumPy uses less memory than Python lists

3. Convenience: NumPy provides many mathematical functions and tools

4. Foundation: Many other data science libraries (Pandas, Matplotlib, scikit-learn) are built on NumPy

```python
# import numpy
import numpy as np

# check the version
version = np.__version__
print(version)
```

# **Create Numpy Arrays**

Arrays are the main data structure in NumPy. An array is a collection of elements of the same data type, and can have any number of dimensions. A one
dimensional array is called a vector, while a two-dimensional array is called a
matrix.

```python
# create array
arr1 = np.array([1, 2, 3, 4, 5])

# from python list
list1 = [1, 2, 3, 4, 5]
arr1 = np.array(list1)
print(f"1d array:{arr1}")

# 2d array from a nested list
list2 = [[1, 2, 3], [4, 5, 6]]
arr2 = np.array(list2)
print(f"2d array:\n{arr2}")
```

### Built-in arrays

```python
# array of zeros: 1d array with 5 zeros, 3x4 array with zeros
arr3 = np.zeros(5)
print(f"zero array: {arr3}")

arr4 = np.zeros((3, 4))
print(f"zero array:\n{arr4}")

# array of ones
arr5 = np.ones(5)
print(f"\n ones array: \n {arr5}")
```

```python
# array with a speicific value: 1d arr with six sevens, 2x2 arr of fives
arr = np.full(6, 7)
print(f"array of sevens: {arr}")
```

```python
# array with evenly spaced values
# values from 0-9
arr = np.arange(10)
print(f"array from 0-9: {arr}")

# values from 0-18, step 2
arr9 = np.arange(0, 20, 2)
print(f"\n array from 0-18, step 2:\n {arr9}")
```

```python
# create identity matrix
id_matrix = np.eye(3)
print(f"identity matrix:\n {id_matrix}")
```

### Random Arrays

```python
# set seed for reproducibility
np.random.seed(42) # ensure the same sequence of random numbers is generated when the code is run
```

```python
# random integers: let's get 10 random integers from 0 to 19
random_ints = np.random.randint(0, 20, 10)
print(f"random integers: {random_ints}")
```

# **Vectorization**

Vectorized computation is a key feature of NumPy that allows for fast computations on arrays. Instead of performing operations on individual elements, NumPy can perform operations on entire arrays at once, which is much faster.

```python
arr1 = np.array([1, 2, 3, 4, 5])
print(arr1 + 5)
```

# **Array Indexing and slicing**

Numpy arrays can be indexed and sliced similarly to python lists

```python
arr = np.array([1, 2, 3, 4, 5])
arr
```

```python
# first item
print(arr[0])
```

```python
# first 3 items
print(arr[:3])
```

```python
# from 2 to 5
print(arr[1:5])
```

```python
# every second item
print(arr[::2])
```

```python
# reversed array
print(arr[::-1])
```

# **Reshaping Arrays**

Changing the shape of array without changing the data

```python
arr = np.arange(12)
print(arr)
print(f"\n shape: {arr.shape}")
```

```python
# reshape to 2d (3x4)
new_arr = arr.reshape(3, 4)
print(new_arr)
print(f"\n shape: {new_arr.shape}")
```

```python
new_arr1 = new_arr.reshape(4, 3)
print(new_arr1)
```

```python
# reshape to 3d
arr_3d = new_arr1.reshape(2, 2, 3)
print(arr_3d)
```

```python
# automatically calculate the number of columns in an array
arr_auto = new_arr.reshape(3, -1)
arr_auto
```

```python
# flatten array
flat_arr = new_arr.flatten()
flat_arr
```
